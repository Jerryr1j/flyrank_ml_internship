## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
import pandas as pd
import numpy as np

# Load the dataset
url = "https://raw.githubusercontent.com/Jerryr1j/flyrank_ml_internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Select numeric features, excluding identifiers and potential raw leak columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
exclude_cols = ['item_id', 'target', 'label', 'refresh_needed', 'trend_direction', 'trend_pct']
feature_cols = [col for col in numeric_cols if col not in exclude_cols]

# Build feature vector with median imputation for missing values
X_features = df[feature_cols].copy()
X_features = X_features.fillna(X_features.median())

print(f"Feature vector successfully built!")
print(f"Total features selected: {len(feature_cols)}")
print(f"Feature matrix shape: {X_features.shape}")
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
# Leakage check: inspect summary statistics and variance to spot abnormal distributions
print("--- Leakage & Variance Audit ---")
variances = X_features.var().sort_values(ascending=False)
print("Top 5 features by variance:")
print(variances.head(5))

# Check for any constant or zero-variance columns
constant_cols = [col for col in X_features.columns if X_features[col].nunique() <= 1]
print(f"\nConstant columns detected (should be 0): {len(constant_cols)}")
```
