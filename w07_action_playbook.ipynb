## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
#The Action Queue: Prioritizing pages that exhibit high search impressions paired with low click-through rates (RULE_HIGH_IMPR_LOW_CTR) or traffic declines (RULE_DECLINING_TRAFFIC).
#Human-Trusted Rationale: Instead of blind automation, each recommended item comes with a clear reason code so the editorial team understands why a particular page was flagged for an update.
import pandas as pd
import numpy as np

# Load dataset and generate action queue sample
url = "https://raw.githubusercontent.com/Jerryr1j/flyrank_ml_internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Simulate ranking/action queue for top actionable items
if 'impressions' in df.columns:
    df['action_priority_score'] = df['impressions'].fillna(0)
    ranked_action_df = df.sort_values(by='action_priority_score', ascending=False).reset_index(drop=True)
else:
    ranked_action_df = df.copy()

print(f"Action playbook queue loaded. Total actionable records: {len(ranked_action_df):,}")
display(ranked_action_df.head(3))
```

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

```python
import os
import pandas as pd

# Ensure the output directory exists
os.makedirs('work/outputs', exist_ok=True)

# Load dataset to generate the final export queue
url = "https://raw.githubusercontent.com/Jerryr1j/flyrank_ml_internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Compute final priority score for export queue
if 'impressions' in df.columns and 'clicks' in df.columns:
    df['final_priority_score'] = df['impressions'].fillna(0) / (df['clicks'].fillna(1) + 1)
else:
    numeric_cols = df.select_dtypes(include=['number']).columns
    df['final_priority_score'] = df[numeric_cols[0]].rank(ascending=False)

# Sort descending to put highest priority items at the top
final_queue = df.sort_values(by='final_priority_score', ascending=False).reset_index(drop=True)

# Export the final action queue to work/outputs/
output_csv_path = 'work/outputs/final_action_queue.csv'
final_queue.head(100).to_csv(output_csv_path, index=False)

print(f"Successfully exported final action queue!")
print(f"File saved to: {output_csv_path}")
display(final_queue[['final_priority_score']].head(3))
```
