## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import pandas as pd
import os

# Load dataset
url = "https://raw.githubusercontent.com/Jerryr1j/flyrank_ml_internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Ensure output directory exists
os.makedirs('work/outputs', exist_ok=True)

# Define a transparent baseline score heuristic using available columns
if 'impressions' in df.columns and 'clicks' in df.columns:
    # Heuristic: Prioritize pages with high impressions relative to low clicks
    df['baseline_score'] = df['impressions'].fillna(0) / (df['clicks'].fillna(1) + 1)
else:
    # Fallback heuristic using available numeric columns
    numeric_cols = df.select_dtypes(include=['number']).columns
    df['baseline_score'] = df[numeric_cols[0]].rank(ascending=False)

# Rank the dataframe descending (highest score = top priority)
ranked_df = df.sort_values(by='baseline_score', ascending=False).reset_index(drop=True)

# Save ranked queue sample to the required work/outputs path
output_path = 'work/outputs/baseline_action_score.csv'
ranked_df.head(100).to_csv(output_path, index=False)

print(f"Successfully generated ranked queue!")
print(f"Saved top items to: {output_path}")
display(ranked_df.head(3))
```
